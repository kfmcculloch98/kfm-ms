```python
NUM_REALS = 250
# initialize blank data frame with columns for parnme and rows for each realization
df = pd.DataFrame(index=range(NUM_REALS), columns=pars.parnme)
df
```

```python
pars.parlbnd = 0.0001
pars.parubnd = 10

for cn,cg in pars.groupby("well"): # current name, current group

    samples = np.random.lognormal(mean=0, sigma=1, size=NUM_REALS) 
    # randomly sample number of realizations with a normal distribution
    # with mean = 1 and std = 5 to generate a lognormal distribution of multipliers

    df.loc[:,cg.parnme] = samples[:,None] * cg.parval1.values[None,:]
    # populate the dataframe with the sampled values multiplied by the original parameter values for each group 
```

```python
from pathlib import Path
import re

import pandas as pd
import matplotlib.pyplot as plt

par_path = Path(r"c:\python\Personal\kfm-ms\codes\master_run\inversion_level_2.0.par.csv")

# Read the wide table
pars = pd.read_csv(par_path)

# First column is realization name
if "real_name" not in pars.columns:
    raise KeyError("Expected a 'real_name' column.")

row = pars.iloc[0] 
real_name = row["real_name"]

pattern = re.compile(r"_r:(\d+)_c:(\d+)_sp:(\d+)")

records = []
for col in pars.columns:
    if col == "real_name":
        continue

    m = pattern.search(col)
    if m:
        r = int(m.group(1))
        c = int(m.group(2))
        sp = int(m.group(3))
        records.append({
            "pname": col,
            "r": r,
            "c": c,
            "sp": sp,
            "well": f"{r}_{c}",
            "value": float(row[col]),
        })

df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No parameter columns matched the expected r/c/sp pattern.")

print(df.head())

fig, ax = plt.subplots(figsize=(12, 5))

for well, group in df.groupby("well"):
    group = group.sort_values("sp")
    ax.plot(group["sp"], group["value"], marker="o", linewidth=1, alpha=0.8, label=well)

ax.set_xlabel("Stress period")
ax.set_ylabel("Parameter value")
ax.set_title(f"Parameter values for par_path: {par_path.name} (realization: {real_name})")
ax.legend(title="well", bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=7)

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import re

import pandas as pd
import matplotlib.pyplot as plt

par_path = Path(r"c:\python\Personal\kfm-ms\codes\master_run\inversion_level_1.0.par.csv")

# Choose the well you want to inspect
target_r = 34
target_c = 54
target_well = f"{target_r}_{target_c}"

# Read the wide table
pars = pd.read_csv(par_path)

if "real_name" not in pars.columns:
    raise KeyError("Expected a 'real_name' column.")

# Regex to extract r, c, sp from column names
pattern = re.compile(r"_r:(\d+)_c:(\d+)_sp:(\d+)")

# Keep only columns for the chosen well
well_cols = []
meta = []

for col in pars.columns:
    if col == "real_name":
        continue

    m = pattern.search(col)
    if m:
        r = int(m.group(1))
        c = int(m.group(2))
        sp = int(m.group(3))

        if r == target_r and c == target_c:
            well_cols.append(col)
            meta.append({
                "pname": col,
                "r": r,
                "c": c,
                "sp": sp,
                "well": f"{r}_{c}",
            })

meta_df = pd.DataFrame(meta)

if meta_df.empty:
    raise ValueError(f"No parameter columns found for well {target_well}")

# Sort columns by stress period
meta_df = meta_df.sort_values("sp").reset_index(drop=True)
well_cols = meta_df["pname"].tolist()

# Build a plotting table: rows = realizations, columns = stress periods
plot_df = pars[["real_name"] + well_cols].copy()

# Rename columns to stress period for easier plotting
rename_map = dict(zip(well_cols, meta_df["sp"].astype(str)))
plot_df = plot_df.rename(columns=rename_map)

# Plot all realizations as faint lines
fig, ax = plt.subplots(figsize=(10, 5))

sp_cols = [str(sp) for sp in meta_df["sp"].tolist()]

for _, row in plot_df.iterrows():
    ax.plot(
        meta_df["sp"],
        [row[col] for col in sp_cols],
        color="tab:blue",
        alpha=0.08,
        linewidth=0.8
    )

ax.set_xlabel("Stress period")
ax.set_ylabel("Parameter value")
ax.set_title(f"All realizations for well {target_well}")
ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```
