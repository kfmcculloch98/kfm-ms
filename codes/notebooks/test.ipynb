```python
import os
import shutil
import numpy as np
import flopy
from multiprocessing import Pool, cpu_count
from sim_utils import worker_task  # This must come from the .py file

if __name__ == "__main__":
    # 1. Clean workspace to avoid 'File in use' errors
    if os.path.exists(workspace):
        shutil.rmtree(workspace, ignore_errors=True)
    os.makedirs(workspace)

    # 2. Re-verify candidate_wells and cp_cells are defined
    n_cand = len(candidate_wells)
    tasks = [(j, wr, wc, cp_cells, workspace, mf6_exe, dt, nper) 
             for j, (wr, wc) in enumerate(candidate_wells)]

    print(f"Starting {cpu_count()} parallel workers...")
    
    # 3. Use the Pool
    with Pool(processes=cpu_count()) as pool:
        results = pool.map(worker_task, tasks)

    # 4. Fill R_full
    R_full = np.zeros((nper, len(cp_cells), n_cand))
    for j, data in results:
        if data is not None:
            R_full[:, :, j] = data
        else:
            print(f"Warning: Simulation {j} failed!")

    print("Success! R_full populated. Now run the Global Solve cell.")
```
